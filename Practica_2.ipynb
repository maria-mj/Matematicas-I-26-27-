#### **De funciones simbólicas a funciones numéricas (`sp.lambdify`)**

En SymPy trabajamos con expresiones simbólicas para calcular derivadas, integrales o límites. Sin embargo, los algoritmos numéricos necesitan evaluar esas fórmulas a gran velocidad sobre números o vectores de NumPy.

La función `sp.lambdify` actúa como puente: toma una expresión simbólica de SymPy y la transforma automáticamente en una función numérica ejecutable con NumPy.

```python
import sympy as sp
import numpy as np
```

```python
# 1. Definimos la variable y la expresión simbólica en SymPy
x = sp.Symbol('x')
f_sim = sp.cos(x) - x
```

```python
# f_sim es una función simbólica; para evaluarla debemos usar .subs
print("La función simbólica es:", f_sim, "y su valor en 0:", f_sim.subs(x,0))
```

```python
# 2. Convertimos la expresión simbólica a función numérica de NumPy
f = sp.lambdify(x, f_sim, "numpy")
```

```python
# Ahora podemos evaluar la función directamente:
print("Podemos calcular f(0) =", f(0))
```

## Bloque 2: Método de la Bisección

El Método de la Bisección se fundamenta en el **Teorema de Bolzano**: si $f(x)$ es continua en $[a, b]$ y $f(a) \cdot f(b) < 0$, existe al menos una raíz $c \in (a, b)$ tal que $f(c) = 0$.

**Algoritmo**

En cada iteración se calcula el punto medio $c = \dfrac{a+b}{2}$ y se selecciona el subintervalo donde cambia el signo de la función.

```python
def biseccion(f, a, b, tol=1e-5, max_iter=50):
    if f(a) * f(b) >= 0:
        print("El Teorema de Bolzano no se cumple en el intervalo dado.")
        return None

    for i in range(max_iter):
        c = (a + b) / 2.0

        if abs(f(c)) < tol or (b - a) / 2.0 < tol:
            print(f"Convergencia alcanzada en {i+1} iteraciones.")
            return c

        if f(a) * f(c) < 0:
            b = c
        else:
            a = c

    print("Alcanzado el número máximo de iteraciones.")
    return c
```

Algunos ejemplos de uso de la función:

```python
# Ejemplo: calcula una raíz de f(x) = x^3 - x - 2 en [1, 2]
f = lambda x: x**3 - x - 2
raiz = biseccion(f, 1.0, 2.0, tol=1e-4)
print("Valor aproximado de la raíz:", raiz)
```

```python
# Ejemplo: calcula una aproximación de sqrt(2) buscando una raíz de x^2-2=0.
f1 = lambda x: x**2 - 2

raiz_sqrt2 = biseccion(f1, 1.0, 2.0)
print("Valor aproximado de la raíz:", raiz_sqrt2)
print("Error respecto al valor exacto:", abs(raiz_sqrt2 - np.sqrt(2)))
```

## Bloque 3: Método de Newton-Raphson (`sympy` + `numpy`)

El Método de Newton-Raphson utiliza la **recta tangente** a la curva en cada iteración para aproximar la raíz con mayor velocidad. Sus iteraciones se calculan con:

$$x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)}$$

**Algoritmo del Método de Newton**

```python
# Función del Método de Newton-Raphson
def newton_raphson(f, df, x0, tol=1e-5, max_iter=50):
    x = x0
    for i in range(max_iter):
        derivada = df(x)
        if abs(derivada) < 1e-12:
            print("Error: Derivada nula o casi nula.")
            return None

        x_nuevo = x - f(x) / derivada
        if abs(x_nuevo - x) < tol:
            print(f"Newton-Raphson convergió en {i+1} iteraciones.")
            return x_nuevo
        x = x_nuevo

    print("Se alcanzó el límite de iteraciones sin convergencia.")
    return x
```

Como para aplicar este método necesitamos trabajar tanto con la función como con su derivada, utilizaremos la librería `Sympy` para calcular estas derivadas de manera directa y la instrucción `sp.lambdify` para transformar estas funciones simbólicas a funciones de `Numpy` que podamos evaluar directamente sobre los puntos necesarios.



```python
# 1. Cargamos los paquetes
import sympy as sp
import numpy as np
```

```python
# 2. Definimos la función de manera simbólica y calculamos y su derivada
x = sp.Symbol('x')

f_sim = sp.exp(-x) - x
df_sim = sp.diff(f_sim, x)
```

```python
# 3. Convertimos las funciones simbólicas a funciones numéricas ejecutables con NumPy
f = sp.lambdify(x, f_sim, "numpy")
df = sp.lambdify(x, df_sim, "numpy")
```

```python
# 4 Calculamos una raíz de f(x)=e^(-x)-x aplicando la función newton_raphson
raiz_newton = newton_raphson(f, df, x0=0)
print("Raíz calculada con Newon-Raphson:", raiz_newton)
```

> ⚠️ **Atención - Limitaciones de Newton-Raphson:**
> 1. **Derivada nula ($f'(x_n) = 0$):** Causa una división por cero interrumpiendo el algoritmo.
> 2. **Elección del valor inicial $x_0$:** Si $x_0$ se elige lejos de la solución o cerca de un extremo local, el método puede diverger o converger a una raíz diferente.

## 🚀 Extra: Comparativa de Convergencia (Bisección vs Newton-Raphson)


El método de la **Bisección** es muy seguro y estable, pero su convergencia es *lineal* (lenta). Por el contrario, el método de **Newton-Raphson** presenta convergencia *cuadrática*, lo que significa que duplica aproximadamente el número de dígitos exactos en cada paso (siempre que $x_0$ esté suficientemente cerca de la raíz).

Comprobemos cuántas iteraciones requiere cada método para hallar una raíz de $f(x) = x^3 - x - 2$:

```python
# Función y derivada simbólica
x_sym = sp.Symbol('x')
expr = x_sym**3 - x_sym - 2
d_expr = sp.diff(expr, x_sym)

# Función y derivada numérica
f_num = sp.lambdify(x_sym, expr, "numpy")
df_num = sp.lambdify(x_sym, d_expr, "numpy")

print("--- Bisección en [1, 2] ---")
r_bis = biseccion(f_num, 1.0, 2.0, tol=1e-6)

print("\n--- Newton-Raphson con x0 = 1.5 ---")
r_newt = newton_raphson(f_num, df_num, x0=1.5, tol=1e-6)
```

## Bloque 2: Operaciones Matemáticas Avanzadas con `sympy`
En este bloque utilizaremos la librería de cálculo simbólico **SymPy** para definir funciones, obtener valores exactos y resolver límites, derivadas e integrales de manera analítica.


```python
# Primero nos aseguramos de tener instalada la librería
!pip install sympy
```

```python
# Importar paquetes de matemáticas
import sympy as sp
```

```python
# Definimos la variable simbólica 'x'
x = sp.Symbol("x")
```

```python
# 1. Definición de una función simbólica f(x) = x^2*sin(x)
f = x**2*sp.sin(x)
print("Función f(x) =", f)
```

```python
# 2. Evaluación exacta y numérica (.subs y .evalf)
val_exacto = f.subs(x, sp.pi / 2)
val_numerico = val_exacto.evalf()

print("f(pi/2) exacto:", val_exacto)
print("f(pi/2) numérico aprox.:", val_numerico)

## Aproximación de decimales (redondeo)
val_aprx = val_exacto.evalf(3)

print("f(pi/2) aproximado a 3 cifras:", val_aprx)
```

```python
# 3. Cálculo de Límites (sp.limit)
g = (1-sp.cos(x))/x**2
limite_0 = sp.limit(g,x,0)

print("Límite de (1-cos(x))/x^2 cuando x->0:", limite_0)

## Límites laterales
lim_der = sp.limit(g,x,0,"+")
lim_izq = sp.limit(g,x,0,"-")

print("Límite cuando x->0+:", lim_der)
print("Límite cuando x->0-:", lim_izq)
```

```python
# 4. Cálculo de Derivadas (sp.diff)
df = sp.diff(f,x)
d5f = sp.diff(f,x,5)

print("Primera derivada f'(x) =", df)
print("Quinta derivada f^(5)(x) =", d5f)
```

```python
# 5. Integrales Indefinidas y Definidas (sp.integrate)
int_indef = sp.integrate(sp.sin(x),x)
int_def = sp.integrate(x**2,(x,0,3))

print("Integral indefinida de sin(x):", int_indef)
print("Integral definida de x^2 entre 0 y 3:", int_def)
```

**Representación gráfica**

Vamos a representar la función $f(x)=x \cos(x)$.

```python
# 1. Creamos una partición de un intervalo del eje x
x = np.arange(0,10,0.1)      # "Partimos" el intervalo [0,10] en subintervalos de longitud 0.1

# 2. Calculamos la imagen de cada punto
y = x*np.cos(x)

# Pintamos el vector de puntos 'x' respecto al vector de imágenes 'y'
plt.plot(x,y)

# Añadimos etiquetas a los ejes
plt.xlabel('Eje x')
plt.ylabel('Eje y')

# Ponemos un título al gráfico
plt.title('Función $f(x)=x·cos(x)$')

# Mostramos el gráfico
plt.show()
```

## 🚀 Extra: ¿Cómo conectar ambos mundos? (De SymPy a NumPy)

Es muy habitual querer lo mejor de ambos mundos:

1.   Modelar o simplificar una **fórmula** compleja algebraicamente con **SymPy**.
2.   **Evaluar** esa fórmula en miles de puntos o graficarla a máxima velocidad con **NumPy**.

Si intentas evaluar una expresión de SymPy punto por punto usando `.subs()`, el proceso será excesivamente lento (y costoso). Para evitarlo, existe la función `sp.lambdify()`, que actua como un traductor: toma una expresión simbólica de SymPy y la convierte en una función nativa de NumPy altamente optimizada.



💻 **Ejemplo**

```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

# -------------------------------------------------------------
# PASO 1: Trabajo simbólico en SymPy
# -------------------------------------------------------------
x = sp.Symbol('x')
# Definimos una función matemática simbólica
f_sympy = sp.cos(x)*sp.exp(-x/5)

print("1. Expresión simbólica (SymPy):")
display(f_sympy)

# -------------------------------------------------------------
# PASO 2: El puente -> sp.lambdify()
# -------------------------------------------------------------
# Convertimos la expresión de SymPy en una función ejecutable por NumPy
# Argumentos: (variable_independiente, expresion_sympy, módulo_destino)
f_numpy = sp.lambdify(x, f_sympy, 'numpy')

# -------------------------------------------------------------
# PASO 3: Cálculo masivo y gráfica en NumPy / Matplotlib
# -------------------------------------------------------------
# Generamos 1000 puntos en un instante
x_vals = np.linspace(0, 20, 1000)

# Evaluamos todos los puntos simultáneamente
y_vals = f_numpy(x_vals)

# Graficamos el resultado
plt.figure(figsize=(8, 4))
plt.plot(x_vals, y_vals, label=r'$f(x) = \cos(x) \cdot e^{-x/5}$', color='tab:purple')
plt.title("Evaluación vectorial ultrarrápida tras usar lambdify")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.legend()
plt.show()
```

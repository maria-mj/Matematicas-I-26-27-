## Bloque 1: Método del Trapecio

El Método del Trapecio aproxima la integral definida $\displaystyle\int_{a}^{b} f(x) \, dx$ sumando las áreas de $n$ trapecios construidos sobre el intervalo $[a, b]$:

$$I_{\text{Trap}} = \frac{h}{2} \left[ f(x_0) + 2 \sum_{i=1}^{n-1} f(x_i) + f(x_n) \right]$$

donde $h = \dfrac{b-a}{n}$ y $x_i = a + i \cdot h$.

```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
```

```python
def trapecio(f, a, b, n):
    """
    Aproxima la integral definida de f(x) en [a, b] usando el Método del Trapecio en n subintervalos.
    """
    h = (b-a)/n
    x = np.linspace(a, b, n+1)
    y = f(x)

    integral = (h/2)*(y[0] + 2*np.sum(y[1:-1]) + y[-1])
    return integral
```

**Ejemplo**

Aproxima el valor de la integral de $f(x) = e^{-x^2}$ entre $[0, 2]$.

```python
# Definimos la función (numérica)
f = lambda x: np.exp(-x**2)

# Calculamos la aproximación con Trapecio
val_trap = trapecio(f, 0, 2, 4)
print("Aproximación por Trapecio:", val_trap)
```

## Bloque 2: Método de Simpson 1/3

El Método de Simpson aproxima la curva ajustando parábolas (polinomios de 2º grado) en cada par de subintervalos. Para poder aplicar este algoritmo, el número de subintervalos $n$ debe ser **par**:

$$I_{\text{Simp}} = \frac{h}{3} \left[ f(x_0) + 4 \sum_{i \text{ impar}}^{n-1} f(x_i) + 2 \sum_{i \text{ par}}^{n-2} f(x_i) + f(x_n) \right]$$

```python
def simpson(f, a, b, n):
    """
    Aproxima la integral definida de f(x) en [a, b] usando el Método de Simpson 1/3.
    Requiere que 'n' sea un número PAR.
    """
    if n % 2 != 0:
        print("Error: El número de subintervalos debe ser un número PAR.")
        return None

    h = (b - a) / n
    x = np.linspace(a, b, n + 1)
    y = f(x)

    # y[1:-1:2] -> selecciona elementos en posiciones impares
    # y[2:-2:2] -> selecciona elementos en posiciones pares intermedias
    integral = (h / 3) * (y[0] + 4 * np.sum(y[1:-1:2]) + 2 * np.sum(y[2:-2:2]) + y[-1])
    return integral
```

> ⚠️ **Atención:**
>
> El Método de Simpson requiere siempre que el número de subintervalos $n$ sea **PAR** ($n = 2, 4, 6, 8, \dots$), ya que cada parábola requiere 3 puntos (2 subintervalos) para su construcción.

**Ejemplo**

Aproxima el valor de la integral de $f(x) = e^{-x^2}$ entre $[0, 2]$.

```python
# Definimos la función (numérica)
f = lambda x: np.exp(-x**2)

# Calculamos la aproximación con Simpson
val_trap = simpson(f, 0, 2, 4)
print("Aproximación por Simpson:", val_trap)
```

## **Comparación de errores**

```python
# Comparación: Valor Exacto (SymPy) vs. Simpson vs. Trapecio
x_sym = sp.Symbol('x')
f_sym = sp.exp(-x_sym**2)
valor_exacto = sp.integrate(f_sym, (x_sym, 0, 2)).evalf()

f_num = lambda x: np.exp(-x**2)
valor_simp = simpson(f_num, 0, 2, n=4)
valor_trap = trapecio(f_num, 0, 2, n=4)

print("Valor Exacto (SymPy): ", valor_exacto)
print("Simpson (n=4): ", valor_simp, "Error: ", abs(valor_exacto - valor_simp))
print("Trapecio (n=4): ", valor_trap, "Error: ", abs(valor_exacto - valor_trap))
```

## Bloque 3: Visualización del Área Integrada con `matplotlib`

Para representar gráficamente el área bajo la curva correspondiente a una integral definida, podemos utilizar la instrucción `plt.fill_between()` de Matplotlib.

```python
x_vals = np.linspace(0, 2, 200)
f_num = lambda x: np.exp(-x**2)

plt.figure(figsize=(8, 4))
plt.plot(x_vals, f_num(x_vals), label="$f(x) = e^{-x^2}$", color="darkblue", linewidth=2)

# Sombrear el área bajo la curva
plt.fill_between(x_vals, f_num(x_vals), color="skyblue", alpha=0.4, label="Área Integrada")

plt.title("Visualización del Área bajo la Curva", fontsize=13)
plt.xlabel("Eje X")
plt.ylabel("Eje Y")
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend()
plt.show()
```
